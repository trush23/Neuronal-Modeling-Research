# 01 · Neuronal model catalog
## Goal
Build a research menu for **single-neuron dynamics first**, then coupled pairs, homogeneous networks, and finally heterogeneous networks. This notebook is a literature-backed catalog, not a simulation study or a ranking by biological realism. It prioritizes normal forms and low-dimensional phenomenological models. No empirical data are used.

**Suggested mathematical spine:** theta/QIF → Hopf normal forms → FitzHugh–Nagumo → slow–fast bursting. Boolean and integrate-and-fire models provide useful comparison baselines.

The directory contains supplied research PDFs; this first catalog uses the linked references below and does **not** claim to have reviewed those PDFs. Model selection, physical calibration, coupling, and graph parameters remain research decisions.

## 1. Vocabulary and conventions
- A **normal form** retains leading terms near a specified bifurcation after coordinate/time changes. Its validity is generally local. A low-dimensional model is not automatically a normal form.
- **Type I excitability:** repetitive firing can begin with arbitrarily small frequency. A SNIC (saddle-node on invariant circle) is a canonical mechanism; other mechanisms exist.
- **Type II excitability:** onset occurs at nonzero limiting frequency; Hopf is a canonical mechanism. Type I/II firing onset should not be conflated with a phase-response-curve classification without checking assumptions.
- **Hybrid models** combine continuous flow with event/reset rules. **Maps/Boolean models** advance by discrete updates; one update is not automatically one millisecond.
- A **phase reduction** describes motion near an existing stable oscillation under weak perturbations. The theta neuron instead captures the SNIC transition, including rest and excitability.

Unless physical units are explicitly given, coordinates, input, and time are dimensionless. A dot denotes derivative with respect to the model's own time. Inputs called $I$ are model-specific and are not numerically interchangeable. Every equation below specifies a convention; signs and scalings vary between publications.

No noise, delay, coupling, or default parameter sets are assumed in this catalog. Reported behaviors are possible regimes, not guaranteed behavior for arbitrary parameters.

## 3. Discrete baselines
### 3.1 Boolean threshold neuron
A modern weighted-threshold convention is
$$s_i[n+1]=H\!\left(\sum_j w_{ij}s_j[n]+b_i[n]-h_i\right),\qquad s_i\in\{0,1\},$$
with $H(x)=1$ for $x\geq0$ and $0$ otherwise. For an isolated cell, replace the sum by prescribed external input. This is inspired by McCulloch–Pitts logic; it is not a literal transcription of their inhibitory-veto construction [R1].

**Use:** discrete activation, attractors, switching, and graph baselines. **Limit:** no membrane trajectory, intrinsic spike waveform, or refractory period. Synchronous versus asynchronous updating can change attractors. Hopfield energy/convergence properties require additional conditions and do not hold for arbitrary Boolean networks.

### 3.2 Excitable cellular automaton
One common Greenberg–Hastings-type convention has $q\geq3$ states: $0$ resting, $1$ active, and $2,\ldots,q-1$ refractory:
$$s_i[n+1]=\begin{cases}
1,&s_i[n]=0\ \text{and}\ \sum_j A_{ij}\mathbf1_{s_j[n]=1}\geq m,\\
0,&s_i[n]=0\ \text{and excitation condition fails},\\
(s_i[n]+1)\bmod q,&s_i[n]>0.
\end{cases}$$
Here $m\geq1$ is an integer neighbor threshold; the update is synchronous. An isolated resting cell needs an external trigger. **Use:** propagation and refractory waves. **Limit:** propagation rules are imposed, not derived from ionic dynamics; lattice spirals depend on rules and conditions [R2]. This is a specified baseline variant, not a unique Boolean neuron.

## 4. Canonical onset models — highest priority
### 4.1 Theta neuron and 4.2 QIF
Choose the dimensionless QIF convention
$$\dot v=v^2+\eta+I(t),\qquad +\infty\mapsto-\infty.$$
With $v=\tan(\theta/2)$, the equivalent circle equation is
$$\dot\theta=1-\cos\theta+(1+\cos\theta)[\eta+I(t)],\qquad\theta\in S^1.$$
A spike is a forward crossing of $\theta=\pi\pmod{2\pi}$. For constant effective drive $a=\eta+I$:
- $a<0$: QIF equilibria $v_\pm=\pm\sqrt{-a}$; $v_-$ is stable and $v_+$ unstable.
- $a=0$: saddle-node threshold; the global circle/reinjection supplies the SNIC structure.
- $a>0$: rotation, with $T=\int_{-\infty}^{\infty}dv/(v^2+a)=\pi/\sqrt a$ and $f=\sqrt a/\pi$ per model-time unit.

**Use:** analytically tractable Type I excitability, onset scaling, pulse response, phase interactions. **Limit:** no independent amplitude/recovery variable or intrinsic adaptation. A finite QIF threshold/reset changes its period; the exact equivalence requires ideal reinjection. Coupling must transform with coordinates: additive $I$ in QIF becomes $(1+\cos\theta)I$ in theta [R3].

### 4.3 Supercritical Hopf / Stuart–Landau oscillator
For $z=x+iy$, choose
$$\dot z=(\mu+i\omega_0)z-(1+ic)|z|^2z+u(t).$$
Without forcing, $\dot r=\mu r-r^3$ and $\dot\phi=\omega_0-cr^2$. The origin is stable for $\mu<0$; a stable cycle $r=\sqrt\mu$ appears for $\mu>0$. Assume $\omega_0\ne0$ and work near $\mu=0$ so angular speed remains nonzero.

**Use:** local oscillation onset, amplitude relaxation, frequency shifts, and coupled-oscillator theory. **Limit:** the emerging oscillation is small; it is not automatically a full action potential. A declared phase crossing or observation threshold is needed to count spikes. Finite-frequency onset alone does not make every Hopf oscillator a realistic excitable neuron [R4].

### 4.4 Subcritical Hopf with quintic saturation
Choose
$$\dot z=(\mu+i\omega_0)z+(\beta+ic_3)|z|^2z-|z|^4z+u(t),\qquad\beta>0.$$
The radial equation is $\dot r=r(\mu+\beta r^2-r^4)$. For $-\beta^2/4<\mu<0$, stable rest coexists with an outer stable cycle, separated by an unstable cycle. The cycle fold is at $\mu=-\beta^2/4$; the origin loses stability at $\mu=0$. Choose $\omega_0+c_3r^2\ne0$ on cycles of interest.

**Use:** bistability, finite-amplitude switching, hysteresis. **Limit:** the quintic term provides saturation beyond the cubic local Hopf form; the outer cycle need not lie in a quantitatively valid local neighborhood. Hysteresis requires a suitable sweep/protocol [R4].

## 5. Smooth excitable dynamics
### 5.1 FitzHugh–Nagumo (FHN)
A common dimensionless convention is
$$\dot v=v-v^3/3-w+I(t),\qquad\dot w=\epsilon(v+a-bw),\quad\epsilon>0.$$
$v$ is the fast activation variable; $w$ is recovery. For $\epsilon\ll1$, nullclines $w=v-v^3/3+I$ and $w=(v+a)/b$ (when $b\ne0$) organize slow–fast motion.

**Use:** rest, excitable excursions, repetitive oscillations, nullclines, and reaction–diffusion patterns. **Limit:** onset and stability depend on parameters; do not assign every FHN convention a single excitability class. It is a phenomenological reduction, not itself one universal bifurcation normal form. Standard autonomous planar FHN cannot support deterministic chaos as a bounded attractor under the usual planar-flow hypotheses. Spike counting requires an explicit section/threshold and direction [R5].

For equilibria under constant input, the Jacobian is
$$J=\begin{pmatrix}1-v_*^2&-1\\\epsilon&-\epsilon b\end{pmatrix}.$$
A candidate Hopf point requires $\operatorname{tr}J=0$, $\det J>0$, and additional nondegeneracy/transversality checks; a trace crossing alone is insufficient.

## 6. Threshold/reset comparison models
### 6.1 Leaky integrate-and-fire (LIF)
$$\tau_m\dot V=-(V-E_L)+RI(t).$$
At $V=V_{th}$ emit a spike, reset to $V_r<V_{th}$, and optionally hold for $t_{ref}$. $V,E_L,V_r,V_{th}$ are voltages; $\tau_m,t_{ref}$ are times; $RI$ is voltage.

**Use:** passive integration and transparent threshold/reset baseline. **Limit:** the threshold/reset is imposed; spike waveforms and active subthreshold resonance are absent. Under constant $I$ with $V_\infty=E_L+RI>V_{th}$,
$$T=t_{ref}+\tau_m\log\frac{V_\infty-V_r}{V_\infty-V_{th}}.$$
Do not identify its onset mechanism with the SNIC solely because frequency can approach zero [R6].

### 6.2 Exponential integrate-and-fire (EIF)
$$C\dot V=-g_L(V-E_L)+g_L\Delta_T\exp[(V-V_T)/\Delta_T]+I(t).$$
A finite numerical cutoff $V_{cut}$ triggers reset to $V_r$. $V_T$ is the exponential onset parameter, not the event cutoff. **Use:** sharp spike initiation. **Limit:** cutoff/reset and convergence must be declared; this is not a global normal form [R7].

### 6.3 Adaptive exponential integrate-and-fire (AdEx)
$$C\dot V=-g_L(V-E_L)+g_L\Delta_T e^{(V-V_T)/\Delta_T}-w+I(t),$$
$$\tau_w\dot w=a(V-E_L)-w.$$
At the cutoff: $V\leftarrow V_r$, $w\leftarrow w+b$. Here $w,b,I$ are currents and $a$ a conductance. **Use:** adaptation and parameter-dependent firing/bursting patterns. **Limit:** requires both flow and event handling; two state variables do not make this a smooth planar flow [R8].

### 6.4 Izhikevich model
In its published numerical convention (time in ms and $v$ in mV),
$$\dot v=0.04v^2+5v+140-u+I,\qquad\dot u=a(bv-u),$$
$$v\geq30\quad\Rightarrow\quad v\leftarrow c,\quad u\leftarrow u+d.$$
The coefficients carry the corresponding implied scaling; $I$ is a model input, not automatically an experimentally calibrated current. **Use:** a compact spiking/bursting repertoire. **Limit:** hybrid reset dynamics; do not call the whole system a universal normal form. Event timing and update ordering affect simulations [R9].

## 7. Bursting extensions
### 7.1 Hindmarsh–Rose (HR)
A standard dimensionless three-state convention is
$$\dot x=y-ax^3+bx^2-z+I,\qquad\dot y=c-dx^2-y,$$
$$\dot z=r[s(x-x_R)-z],\qquad0<r\ll1.$$
$x$ is voltage-like, $y$ fast recovery, and $z$ slow adaptation. **Use:** tonic spiking, bursting, and parameter-dependent irregular/chaotic dynamics. **Limit:** chaos is not established by an irregular trace; it needs suitable Lyapunov/convergence diagnostics. A slow variable enables burst modulation but does not guarantee bursting. This phenomenological model is not a single onset normal form [R10].

### 7.2 Rulkov spiking map — explicitly the 2002 piecewise variant
$$x_{n+1}=F_\alpha(x_n,y_n),\qquad y_{n+1}=y_n-\varepsilon(x_n+1-\sigma),\quad0<\varepsilon\ll1,$$
$$F_\alpha(x,y)=\begin{cases}
\alpha/(1-x)+y,&x\leq0,\\
\alpha+y,&0<x<\alpha+y,\\
-1,&x\geq\alpha+y.
\end{cases}$$
Use $\alpha>0$ and give the first branch precedence when $x\leq0$; the last branch applies to the remaining positive-$x$ cases. There is no external forcing in this displayed autonomous variant.

**Use:** discrete-time spiking/bursting at low computational cost. **Limit:** update scale is not physical time until calibrated; maps can exhibit chaos even in two dimensions. Several distinct models are called “Rulkov map”; do not interchange this piecewise map with the earlier $\alpha/(1+x^2)+y$ variant or reuse parameters between them [R11].

## 8. Phase reduction — for later synchronization studies
For a stable limit-cycle neuron under weak input, a first-order reduction has the form
$$\dot\phi_i=\omega_i+Z_i(\phi_i)p_i(t),$$
where $Z_i$ is the phase response curve for the specified perturbation channel. Averaging weak interactions can yield
$$\dot\phi_i=\omega_i+K\sum_j A_{ij}H_{ij}(\phi_j-\phi_i).$$
The Kuramoto sine interaction is one special choice, not a generic synaptic law. **Use:** phase locking, synchrony, and graph effects. **Limit:** loses amplitude, adaptation, rest-to-spiking transitions, and strong-coupling effects; validity requires checking reduction assumptions. Derive or estimate $Z$ and $H$ for the chosen parent model [R12].

**Biophysical boundary:** Morris–Lecar and Hodgkin–Huxley can later serve as conductance-based reference models. They are outside the main menu because the present priority is normal-form/phenomenological dynamics. Wilson–Cowan describes interacting population activities, so it should not silently be labeled a single-neuron model.

## 10. Checks — analytic benchmarks for later implementations
The following checks evaluate formulas, not numerical trajectories. They establish small reference cases for future solvers. Frequencies below are in inverse dimensionless time, **not Hz**.

```python
# Theta/QIF transformation identity, away from the tangent singularity.
for theta in (-2.0, -0.7, 0.0, 1.4, 2.8):
    drive = 0.25
    v = math.tan(theta / 2)
    transformed = 2 * (v*v + drive) / (1 + v*v)
    theta_rhs = 1 - math.cos(theta) + drive * (1 + math.cos(theta))
    assert math.isclose(transformed, theta_rhs, rel_tol=1e-12, abs_tol=1e-12)

# Radial equilibria and radial stability for the quintic Hopf model.
beta, mu = 1.0, -0.1
q_inner = (beta - math.sqrt(beta*beta + 4*mu)) / 2
q_outer = (beta + math.sqrt(beta*beta + 4*mu)) / 2
for q in (q_inner, q_outer):
    assert math.isclose(mu + beta*q - q*q, 0.0, abs_tol=1e-12)
assert mu < 0  # stable origin
assert 2*q_inner*(beta - 2*q_inner) > 0  # unstable inner cycle
assert 2*q_outer*(beta - 2*q_outer) < 0  # stable outer cycle
assert len({m['id'] for m in MODEL_MENU}) == len(MODEL_MENU)

print('Analytic identity and radial-stability checks passed.')
for drive in (0.01, 0.25, 1.0):
    print(f'QIF drive={drive:.2f}: period={math.pi/math.sqrt(drive):.6f}, '
          f'frequency={math.sqrt(drive)/math.pi:.6f}')
```

### Future coupling contract
For a smooth node, a candidate template is
$$\dot{\mathbf x}_i=F_{m_i}(\mathbf x_i;p_i)+B_i(\mathbf x_i)I_i^{coupling}.$$
This does not directly specify a Boolean/map/hybrid implementation. Plausible candidate channels include:
- Electrical coupling between calibrated voltage variables: $I_i=\sum_j g_{ij}(V_j-V_i)$, with conductance/current units and typically symmetric nonnegative gap-junction conductances.
- Chemical synapses: $I_i=\sum_j g_{ij}s_j(t)(E_{rev,j}-V_i)$ under the inward-current-positive convention; synaptic kinetics, reversal potentials, and spike rules must be specified.
- Phase coupling: reduced interactions justified by the chosen parent dynamics.

A dimensionless phase, Boolean state, and voltage cannot be coupled by direct subtraction without an explicit observation/input map and calibration. Mixed continuous/discrete models also need a declared scheduling rule.

### Future graph controls
| Graph family | Specification required |
|---|---|
| Square lattice | Side lengths, 4/8 neighbors, open/periodic boundaries |
| Complete | Directed/undirected, self-loop policy |
| Complete bipartite / general bipartite | Partition sizes, edge rule and direction; partitions need not mean excitation/inhibition |
| Erdős–Rényi | $G(N,p)$ or $G(N,M)$, seed, directedness |
| Random regular | Degree and existence constraints |
| Sparse | An edge-density/degree regime, not a unique graph ensemble |
| Small-world | Base ring degree, rewiring rule/probability |
| Preferential attachment | Attachment rule, seed graph, edges per added node |

Compare fixed per-edge strength and degree-normalized input as distinct experiments: they answer different questions. Keep $N$, density/degree, connectivity, weights, direction, delays, and graph seeds explicit; use ensembles to quantify variability. Collective observables may include firing-rate distributions, phase order (only with meaningful phases), spike/burst synchrony, waves, and multistability. These are planned measurements, not established outcomes.
